## Simple Queue

```python
class SQ():
  def __init__(self, maxsize) -> None:
    self.front = 0
    self.rear = 0
    self.maxsize = maxsize
    self.list: List[int | None] = [None for _ in range(maxsize)]

  def push(self, item):
    if self.rear >= self.maxsize:
      raise Exception("Queue is full")

    self.list[self.rear] = item
    self.rear += 1

  def pop(self):
    if self.front > self.rear:
      raise Exception("Queue is empty")
    
    self.list[self.front] = None
    self.front += 1

    if self.front >= self.rear:
      self.list = [None for _ in range(self.maxsize)]
      self.rear = self.front = 0

  def __repr__(self) -> str:
    return ",".join(map(str, self.list[:self.rear]))
```

## Circular Queue

```python
class CQ():
  def __init__(self, maxsize) -> None:
    self.front = None
    self.rear = None
    self.maxsize = maxsize
    self.list: List[int | None] = [None for _ in range(maxsize)]

  def push(self, item):
    if self.rear == None:
      self.rear = 0
      self.front = 0
    elif self.front == (self.rear + 1) % self.maxsize:
      raise Exception("Queue is full")
    else:
      self.rear = (self.rear + 1) % self.maxsize

    self.list[self.rear] = item

  def pop(self):
    if self.front == None:
      raise Exception("Queue is empty")
    
    value = self.list[self.front]
    self.list[self.front] = None
    self.front = (self.front + 1) % self.maxsize

    if self.rear and self.front == (self.rear + 1) % self.maxsize:
      self.rear = 0
      self.front = 0

    return value

  def __repr__(self) -> str:
    return ",".join(map(str, self.list))
```

```python
class Heap():
  def min_order(x,y):
    return x < y
  
  def max_order(x,y):
    return x > y
  
  def __init__(self, ordering=min_order) -> None:
    self.tree = []
    self.order = ordering
    self.capacity = 0
    self.curr_parent = None

  def size(self):
    return len(self.tree)

  def heapify(self, index):
    while True:
      parent = self.get_parent(index)
      if parent == None:
        break

      if not self.order(self.tree[parent], self.tree[index]):
        self.tree[parent], self.tree[index] = self.tree[index], self.tree[parent]
        index = parent
      else:
        break

  def heapify_root(self):
    index = 0

    while True:
      l_index, r_index = self.get_left_child_idx(index), self.get_right_child_idx(index)
      best = index

      if l_index and self.tree[l_index] and not self.order(self.tree[best], self.tree[l_index]):
        best = l_index
      if r_index and self.tree[r_index] and not self.order(self.tree[best], self.tree[r_index]):
        best = r_index
      if best == index:
        break

      self.tree[index], self.tree[best] = self.tree[best], self.tree[index]
      index = best

  def get_left_child_idx(self, index) -> "int | None":
    l_index = 2 * index + 1
    if self.size() - 1 >= l_index:
      return l_index
    return None
  
  def get_right_child_idx(self, index) -> "int | None":
    r_index = 2 * index + 2
    if self.size() - 1 >= r_index:
      return r_index
    return None
  
  def set_left_child(self, index, val):
    l_index = self.get_left_child_idx(index)
    if l_index:
      self.tree[l_index] = val
  
  def set_right_child(self, index, val):
    r_index = self.get_right_child_idx(index)
    if r_index:
      self.tree[r_index] = val

  def get_parent(self, index) -> "int | None":
    if index == 0:
      return None
    elif index % 2 == 0:
      return (index - 2) // 2
    return (index - 1) // 2

  def get_height(self):
    return math.ceil(math.log2(self.size() + 1))

  def extend_tree(self):
    self.tree.extend([None for _ in range(2**(self.get_height()))])

  def push(self, val):
    if self.size() == 0:
      self.tree.append(val)
      self.curr_parent = 0
      self.capacity += 1
      return
    
    if self.size() == self.capacity:
      self.extend_tree()

    while True:
      curr_l_child_idx = self.get_left_child_idx(self.curr_parent)
      curr_r_child_idx = self.get_right_child_idx(self.curr_parent)

      if curr_l_child_idx != None and self.tree[curr_l_child_idx] == None:
        self.set_left_child(self.curr_parent, val)
        break
      elif curr_r_child_idx != None and self.tree[curr_r_child_idx] == None:
        self.set_right_child(self.curr_parent, val)
        break
      else:
        self.curr_parent += 1 # type: ignore

    self.tree[self.capacity] = val
    self.heapify(self.capacity)
    self.capacity += 1

  def pop(self):
    if self.size() == 0:
      raise Exception("Heap is empty")

    latest_child = None

    while latest_child == None:
      curr_l_child_idx = self.get_left_child_idx(self.curr_parent)
      curr_r_child_idx = self.get_right_child_idx(self.curr_parent)

      if curr_r_child_idx != None and self.tree[curr_r_child_idx] != None:
        latest_child = curr_r_child_idx
      elif curr_l_child_idx != None and self.tree[curr_l_child_idx] != None:
        latest_child = curr_l_child_idx
      else:
        self.curr_parent -= 1 # type: ignore

    val = self.tree[0]
    self.tree[0], self.tree[latest_child] = self.tree[latest_child], self.tree[0]  # type: ignore
    self.tree[latest_child] = None # type: ignore
    self.heapify_root()
    self.capacity -= 1
    return val

  def __repr__(self) -> str:
    output = ""
    height = 1
    # i,j = 0,0

    while height <= self.get_height():
      level = []
      for i in range(2**(height-1), 2**height):
        if self.tree[i-1] != None:
          level.append(self.tree[i-1])
      output += f"{','.join(map(str, level))}\n"
      height += 1

    return output
```

# Linked List

```python
class SLLNode():
  def __init__(self, data) -> None:
    self.data = data
    self.next: SLLNode | None = None

class SLL():
  def __init__(self) -> None:
    self.head: SLLNode | None = None
    self.tail: SLLNode | None = None
    self.size = 0

  def insert_at_beg(self, data) -> None:
    node = SLLNode(data)

    if self.head is None:
      self.head = node
      self.tail = node
    else:
      old_head = self.head
      self.head = node
      self.head.next = old_head

    self.size += 1

  def insert_at(self, data, pos) -> None:
    if pos == 0:
      self.insert_at_beg(data)
      return
    elif pos == self.size - 1:
      self.append(data)
      return

    node = SLLNode(data)

    curr_node = self.head
    prev_node: SLLNode | None = None
    index = 0
    while curr_node is not None:
      if index == pos and prev_node:
        prev_node.next = node
        node.next = curr_node
        break

      index += 1
      prev_node = curr_node
      curr_node = curr_node.next

    self.size += 1

  def append(self, data) -> None:
    node = SLLNode(data)
    
    if self.head is None:
      self.head = node
    elif self.tail is not None:
      self.tail.next = node

    self.tail = node
    self.size += 1

  def delete_head(self) -> None:
    if self.head:
      self.head = self.head.next
      self.size -= 1

  def delete_tail(self) -> None:
    self.delete_at(self.size - 1)

  def delete_at(self, pos) -> None:
    if pos == 0:
      self.delete_head()
      return

    curr_node = self.head
    prev_node: SLLNode | None = None
    index = 0

    while curr_node is not None:
      if pos == index and prev_node:
        prev_node.next = curr_node.next

      index += 1
      prev_node = curr_node
      curr_node = curr_node.next

    self.size -= 1

  def __repr__(self) -> str:
    repr = ""
    curr_node = self.head

    while curr_node is not None:
      repr += f"{curr_node.data}, "
      curr_node = curr_node.next

    return "[" + repr.strip()[:-1] + "]"
```
